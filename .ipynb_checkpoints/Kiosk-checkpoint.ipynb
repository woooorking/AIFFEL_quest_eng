# 키오스크 만들기

```python
class Kiosk:
    def __init__(self):
        self.menu = menu
        self.price = price
        self.price_sum = 0

        self.order_menu = []
        self.order_price = []

    # 메뉴 출력 메서드
    def menu_print(self):
        for i in range (len(self.menu)):
            print(i+1, self.menu[i], ':', self.price[i], '원')


    # 주문 메서드
    def menu_select(self):
        print()
        n = int(input("음료 번호를 입력하세요 : "))

        while n !=0: #결제 (0)을 누르기 전까지 계속 반복한다

            if n >0 and n<= len(self.menu): #메뉴에 있는 번호를 잘 말하면
                # 온도 물어보기
                t = 0
                while t !=1 and t !=2: 
                    t = int(input("hot 음료는 1, ice 음료는 2를 입력하세요"))
        
                    # if - else 결정하기! 1이면 무조건 hot, 아니면 ice
                    if t == 1:
                        self.temp = 'hot'
                    else:
                        self.temp = 'ice'

                    #장바구니 담기
                    self.price_sum = self.price_sum + self.price[n-1]
                    self.order_menu.append(f"{self.temp} {self.menu[n-1]}")
                    self.order_price.append(self.price[n-1])
                
                #이상한 번호를 입력했을때
            else:
                print("다시 입력해주세요")
                    
            n = int(input("추가 주문은 음료 번호, 결제는 0을 누르세요 : "))


                      
    # 지불
    def pay(self):
        print(f"결제금액 {self.price_sum}원 입니다.")

        pay_type = input("결제 수단 선택 (1.현금, 2.카드): ")
        if pay_type == '1':
            print('직원을 불러주세요.')
        else:
            print('카드를 꽂아주세요.')


    # 주문서 출력    

    def table(self):
        # 외곽
        print('⟝' + '-' * 30 + '⟞')
        for i in range(3):
            print('|' + ' ' * 31 + '|')

        # 주문 상품명 : 해당 금액
        for i in range(len(self.order_menu)):
            print(f"{self.order_menu[i]}, :, {self.order_price[i]}")
        print(f"총 합계 : {self.price_sum}")

        # 외곽
        for i in range(3):
            print('|' + ' ' * 31+ '|')
        print('⟝' + '-' * 30 + '⟞')
```

```python
a = Kiosk()
a.menu_print()
a.menu_select()
a.table()
```

# 사각형 넓이 구하기

```python
class Square:
    def __init__(self):
        self.square = int(input('넓이를 구하고 싶은 사각형의 숫자를 써주세요.\n 1.평행사변형 2.사다리꼴'))

        if self.square == 1:
            print(self.par())
        
        elif self.square == 2:
            print(self.trape())
        
        else:
            print('1, 2 중에서 다시 입력해주세요')

    # 평행사변형 메서드
    def par(self):
        width, vertical = map(int, input('가로, 세로를 입력하세요').split(','))
        area = width * vertical
        result = '평행사변형의 넓이는 : ' + str(area)
        return result

    # 사다리꼴 메서드
    def trape(self):
        upper,lower,height = map(int, input('윗변, 아랫변, 높이를 입력하세요.').split(','))
        # 사다리꼴 공식: (윗변 + 아랫변) * 높이 / 2
        area = (upper + lower) * height /2
        result = '사다리꼴의 넓이는' + str(area)
        return result
```

```python
a = Square()
```
